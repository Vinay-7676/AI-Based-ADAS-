```python
!pip -q install ultralytics opencv-python-headless

import cv2
import numpy as np
from ultralytics import YOLO
from google.colab import files
from IPython.display import display, Video

model = YOLO("yolov8n.pt")

uploaded = files.upload()
video_path = list(uploaded.keys())[0]

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise Exception("Unable to open video")

fps = cap.get(cv2.CAP_PROP_FPS)
if not fps or fps <= 0:
    fps = 25

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

output_path = "/content/ADAS_Output.mp4"

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

def detect_lanes(frame):

    h, w = frame.shape[:2]

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(gray, (5, 5), 0)
    edges = cv2.Canny(blur, 50, 150)

    mask = np.zeros_like(edges)

    polygon = np.array([[
        (int(w * 0.08), h),
        (int(w * 0.42), int(h * 0.55)),
        (int(w * 0.58), int(h * 0.55)),
        (int(w * 0.95), h)
    ]], dtype=np.int32)

    cv2.fillPoly(mask, polygon, 255)

    cropped = cv2.bitwise_and(edges, mask)

    lines = cv2.HoughLinesP(
        cropped, 1, np.pi / 180, 35,
        minLineLength=30, maxLineGap=100
    )

    overlay = frame.copy()

    if lines is not None:

        left_lines = []
        right_lines = []

        for line in lines:
            x1, y1, x2, y2 = line[0]

            if x2 == x1:
                continue

            slope = (y2 - y1) / (x2 - x1)

            if abs(slope) < 0.35:
                continue

            if slope < 0:
                left_lines.append((x1, y1, x2, y2))
            else:
                right_lines.append((x1, y1, x2, y2))

        for group in [left_lines, right_lines]:

            if not group:
                continue

            xs = []
            ys = []

            for x1, y1, x2, y2 in group:
                xs.extend([x1, x2])
                ys.extend([y1, y2])

            slope, intercept = np.polyfit(ys, xs, 1)

            y_bottom = h
            y_top = int(h * 0.58)

            x_bottom = int(slope * y_bottom + intercept)
            x_top = int(slope * y_top + intercept)

            x_bottom = np.clip(x_bottom, 0, w - 1)
            x_top = np.clip(x_top, 0, w - 1)

            cv2.line(
                overlay,
                (int(x_bottom), y_bottom),
                (int(x_top), y_top),
                (0, 255, 0),
                7
            )

    return cv2.addWeighted(frame, 0.8, overlay, 0.2, 0)


frame_count = 0
warning_count = 0

while cap.isOpened():

    ret, frame = cap.read()

    if not ret:
        break

    frame_count += 1

    frame = detect_lanes(frame)

    results = model(
        frame,
        conf=0.35,
        classes=[0, 2, 3, 5, 7],
        verbose=False
    )

    detections = []

    for result in results:

        for box in result.boxes:

            cls = int(box.cls[0])
            label = model.names[cls]
            confidence = float(box.conf[0])

            x1, y1, x2, y2 = map(
                int, box.xyxy[0].tolist()
            )

            detections.append({
                "label": label,
                "box": (x1, y1, x2, y2)
            })

            cv2.rectangle(
                frame,
                (x1, y1),
                (x2, y2),
                (255, 0, 0),
                2
            )

            cv2.putText(
                frame,
                f"{label} {confidence:.2f}",
                (x1, max(20, y1 - 10)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (255, 255, 0),
                2
            )

    warning = False
    image_area = height * width

    for obj in detections:

        x1, y1, x2, y2 = obj["box"]

        box_area = (x2 - x1) * (y2 - y1)
        center_x = (x1 + x2) // 2

        if (
            box_area / image_area > 0.075
            and width * 0.30 < center_x < width * 0.70
        ):
            warning = True

    if warning:

        warning_count += 1

        cv2.rectangle(
            frame, (0, 0), (width, 55), (0, 0, 200), -1
        )

        cv2.putText(
            frame,
            "WARNING: POSSIBLE FORWARD HAZARD!",
            (10, 35),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.75,
            (255, 255, 255),
            2
        )

    else:

        cv2.putText(
            frame,
            "ADAS STATUS: NORMAL",
            (10, 35),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.75,
            (0, 255, 0),
            2
        )

    cv2.putText(
        frame,
        f"Objects Detected: {len(detections)}",
        (10, height - 45),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (255, 255, 255),
        2
    )

    cv2.putText(
        frame,
        f"Frame: {frame_count}",
        (10, height - 15),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (255, 255, 255),
        2
    )

    out.write(frame)

cap.release()
out.release()

print("ADAS PROCESSING COMPLETED")
print("Total Frames:", frame_count)
print("Warning Frames:", warning_count)

display(Video(output_path, embed=True))

files.download(output_path)
```
